```python
#4. Maior sequência de crescimento
numeros = []
for i in range (10):
  numeros.append (int (input("Digite um número:")))
maior = 1
atual = 1
for i in range (1, len(numeros)):
  if numeros [i] > numeros[i -1]:
    atual +=1
    if atual > maior:
      maior = atual
    else:
      atual = 1
print ("Maior sequência crescente tem tamanho:",maior)
```

```python
#5. Pontos no plano9
pontos = []
for i in range(6):
    x = int(input(f"Digite x{i+1}: "))
    y = int(input(f"Digite y{i+1}: "))
    pontos.append((x, y))

q1 = q2 = q3 = q4 = eixo = origem = 0
maior_dist = 0
ponto_distante = None

for (x, y) in pontos:
    if x == 0 and y == 0:
        origem += 1
    elif x == 0 or y == 0:
        eixo += 1
    elif x > 0 and y > 0:
        q1 += 1
    elif x < 0 and y > 0:
        q2 += 1
    elif x < 0 and y < 0:
        q3 += 1
    else:
        q4 += 1

    dist2 = x**2 + y**2
    if dist2 > maior_dist:
        maior_dist = dist2
        ponto_distante = (x, y)

print("Quadrante I:", q1)
print("Quadrante II:", q2)
print("Quadrante III:", q3)
print("Quadrante IV:", q4)
print("Sobre os eixos:", eixo)
print("Na origem:", origem)
print("Ponto mais distante:", ponto_distante, "com x²+y² =", maior_dist)
```

```python
#6. Frequência e palavra mais repetida
palavras = []
for i in range (10):
  palavra = str (input("Digite uma palavra:"))
  palavras.append(palavra)

frequencia = {}
for p in palavras:
  if p in frequencia:
    frequencia[p] += 1
  else:
    frequencia[p] = 1

maior_freq = 0
for valor in frequencia.values():
  if valor > maior_freq:
    maior_freq = valor

mais_repetidas  = []
for chave, valor in frequencia.items ():
  if valor == maior_freq:
    mais_repetidas.append (chave)
print ("Frequência:",frequencia)
print ("Maior freqûencia:", maior_freq)
print ("Palavras mais repitidas:",mais_repetidas)
```

```python
#7. Boletim de uma turma
turma = []

for i in range(6):
    nome = input(f"Nome do aluno {i+1}: ")
    soma = 0
    for j in range(3):
        nota = float(input(f"Nota {j+1}: "))
        soma += nota
    media = soma / 3
    turma.append({"nome": nome, "media": media})

print("\nBoletim da turma:")
for aluno in turma:
    print(aluno["nome"], "-", f"{aluno['media']:.2f}")

# média geral
soma_medias = 0
for aluno in turma:
    soma_medias += aluno["media"]
media_geral = soma_medias / len(turma)
print("\nMédia geral:", f"{media_geral:.2f}")

# alunos acima da média
acima = 0
for aluno in turma:
    if aluno["media"] >= media_geral:
        acima += 1
print("Alunos com média >= geral:", acima)

# maior média
maior = turma[0]["media"]
for aluno in turma:
    if aluno["media"] > maior:
        maior = aluno["media"]

print("\nAluno(s) com maior média:")
for aluno in turma:
    if aluno["media"] == maior:
        print(aluno["nome"], "-", f"{aluno['media']:.2f}")
```

```python
#8. Controle de estoque com movimentações
# Cadastro dos produtos
estoque = []
for i in range(5):
    codigo = input(f"Código do produto {i+1}: ")
    nome = input("Nome: ")
    preco = float(input("Preço: "))
    quantidade = int(input("Quantidade: "))
    estoque.append({"codigo": codigo, "nome": nome, "preco": preco, "quantidade": quantidade})

# Operações de entrada (E) e saída (S)
for i in range(5):
    op = input("\nOperação (E/S): ").upper()
    codigo = input("Código do produto: ")
    qtd = int(input("Quantidade: "))

    # localizar produto
    encontrado = False
    for p in estoque:
        if p["codigo"] == codigo and qtd > 0:
            encontrado = True
            if op == "E":
                p["quantidade"] += qtd
            elif op == "S":
                if p["quantidade"] >= qtd:
                    p["quantidade"] -= qtd
                else:
                    print("Estoque insuficiente.")
            else:
                print("Operação inválida.")
            break
    if not encontrado:
        print("Código ou quantidade inválida.")

# Mostrar estoque atualizado
print("\nEstoque atualizado:")
valor_total = 0
for p in estoque:
    print(p["codigo"], "-", p["nome"], "-", p["quantidade"], "unidades - R$", f"{p['preco']:.2f}")
    valor_total += p["preco"] * p["quantidade"]

print("\nValor financeiro total do estoque: R$", f"{valor_total:.2f}")
```

```python
#9. Ranking de jogadores
jogadores = []

for i in range(5):
    nome = input(f"Nome do jogador {i+1}: ")
    pontuacoes = []
    for j in range(3):
        pontos = int(input(f"Pontuação da partida {j+1} de {nome}: "))
        pontuacoes.append(pontos)
    jogadores.append({"nome": nome, "pontuacoes": pontuacoes})

# Calcular total e média
for j in jogadores:
    soma = 0
    for p in j["pontuacoes"]:
        soma += p
    j["total"] = soma
    j["media"] = soma / 3

# Mostrar totais e médias
print("\nRanking de jogadores:")
for j in jogadores:
    print(j["nome"], "- Total:", j["total"], "- Média:", f"{j['media']:.2f}")

# Jogador(es) com maior total (sem max)
maior_total = jogadores[0]["total"]
for j in jogadores:
    if j["total"] > maior_total:
        maior_total = j["total"]

print("\nJogador(es) com maior total:")
for j in jogadores:
    if j["total"] == maior_total:
        print(j["nome"], "-", j["total"])

# Jogador(es) mais regulares (menor diferença entre maior e menor pontuação)
menor_dif = jogadores[0]["pontuacoes"][0] - jogadores[0]["pontuacoes"][0]  # começa com 0
for j in jogadores:
    maior = j["pontuacoes"][0]
    menor = j["pontuacoes"][0]
    for p in j["pontuacoes"]:
        if p > maior:
            maior = p
        if p < menor:
            menor = p
    dif = maior - menor
    j["dif"] = dif
    if menor_dif == 0 or dif < menor_dif:
        menor_dif = dif

print("\nJogador(es) mais regulares:")
for j in jogadores:
    if j["dif"] == menor_dif:
        print(j["nome"], "- Diferença:", j["dif"])
```

```python
#10. Agenda de contatos sem duplicidade
agenda = []

# Cadastro dos contatos
while True:
    nome = input("Digite o nome do contato (ou FIM para encerrar): ")
    if nome.upper() == "FIM":
        break
    telefone = input("Digite o telefone: ")

    # Verificar se já existe contato com o mesmo nome
    existe = False
    for contato in agenda:
        if contato["nome"].lower() == nome.lower():
            existe = True
            break

    if existe:
        print("Contato já existe, não será cadastrado.")
    else:
        agenda.append({"nome": nome, "telefone": telefone})

# Consultas
print("\nConsulta de contatos:")
while True:
    nome = input("Digite o nome para consultar (ou FIM para encerrar): ")
    if nome.upper() == "FIM":
        break

    encontrado = False
    for contato in agenda:
        if contato["nome"].lower() == nome.lower():
            print("Telefone de", contato["nome"], ":", contato["telefone"])
            encontrado = True
            break

    if not encontrado:
        print("Contato não encontrado.")
```

```python
#11. Pesquisa de votos
votos = {1: 0, 2: 0, 3: 0, 4: 0}

print("Digite os votos (1 a 4). Digite 0 para encerrar.")

while True:
    codigo = int(input("Voto: "))
    if codigo == 0:
        break
    elif codigo in votos:
        votos[codigo] += 1
    else:
        print("Código inválido.")

# Mostrar totais
print("\nResultado da votação:")
total_votos = 0
for opcao in votos:
    total_votos += votos[opcao]

for opcao in votos:
    if total_votos > 0:
        percentual = (votos[opcao] / total_votos) * 100
    else:
        percentual = 0
    print(f"Opção {opcao}: {votos[opcao]} votos ({percentual:.2f}%)")

# Descobrir vencedor(es) sem max()
maior = votos[1]
for opcao in votos:
    if votos[opcao] > maior:
        maior = votos[opcao]

print("\nOpção(ões) vencedora(s):")
for opcao in votos:
    if votos[opcao] == maior:
        print(f"Opção {opcao} com {votos[opcao]} votos")
```

```python
#12. Histórico de movimentações
estoque = {
    "P01": {"nome": "Arroz", "quantidade": 10},
    "P02": {"nome": "Feijão", "quantidade": 8},
    "P03": {"nome": "Macarrão", "quantidade": 15},
}

historico = []

movimentacoes = [
    ("P01", "E", 5),
    ("P02", "S", 3),
    ("P03", "S", 20),
    ("P01", "S", 2),
    ("P02", "E", 4),
    ("P03", "E", 10),
    ("P01", "S", 20),
    ("P02", "S", 5),
]

for codigo, tipo, qtd in movimentacoes:
    if codigo in estoque and qtd > 0:
        if tipo == "E":
            estoque[codigo]["quantidade"] += qtd
            historico.append((codigo, tipo, qtd))
        elif tipo == "S":
            if estoque[codigo]["quantidade"] >= qtd:
                estoque[codigo]["quantidade"] -= qtd
                historico.append((codigo, tipo, qtd))
            else:
                print("Movimentação recusada:", codigo)
                historico.append((codigo, tipo, qtd, "RECUSADA"))
    else:
        print("Código ou quantidade inválida:", codigo)
        historico.append((codigo, tipo, qtd, "RECUSADA"))

print("\nEstoque final:")
for codigo, dados in estoque.items():
    print(codigo, "-", dados["nome"], "-", dados["quantidade"], "unidades")

print("\nHistórico de movimentações:")
for mov in historico:
    print(mov)
```

```python
#13. Análise de vendas por vendedor
vendedores = []

for i in range(5):
    nome = input(f"Nome do vendedor {i+1}: ")
    vendas = []
    for j in range(4):
        valor = float(input(f"Valor da venda {j+1} de {nome}: "))
        vendas.append(valor)
    vendedores.append({"nome": nome, "vendas": vendas})

for v in vendedores:
    total = 0
    for valor in v["vendas"]:
        total += valor
    v["total"] = total

print("\nRelatório de vendas:")
total_geral = 0
for v in vendedores:
    print(v["nome"], "- Total vendido:", v["total"])
    total_geral += v["total"]

media_equipe = total_geral / len(vendedores)
print("\nTotal geral:", total_geral)
print("Média da equipe:", media_equipe)

maior_total = vendedores[0]["total"]
for v in vendedores:
    if v["total"] > maior_total:
        maior_total = v["total"]

print("\nVendedor(es) com maior total:")
for v in vendedores:
    if v["total"] == maior_total:
        print(v["nome"], "-", v["total"])

acima_media = 0
for v in vendedores:
    if v["total"] > media_equipe:
        acima_media += 1

print("\nQuantidade de vendedores acima da média:", acima_media)
```

```python
#14. Catálogo e pedidos
produtos = [
    {"codigo": "P01", "nome": "Arroz", "preco": 5.50, "estoque": 10},
    {"codigo": "P02", "nome": "Feijão", "preco": 7.20, "estoque": 8},
    {"codigo": "P03", "nome": "Macarrão", "preco": 4.00, "estoque": 15},
    {"codigo": "P04", "nome": "Óleo", "preco": 6.30, "estoque": 12},
    {"codigo": "P05", "nome": "Sal", "preco": 2.50, "estoque": 20},
]

pedidos = [
    ("P01", 5),
    ("P02", 3),
    ("P03", 20),
    ("P04", 2),
    ("P05", 10),
    ("P01", 7),
]

valor_total = 0

print("Processamento dos pedidos:\n")
for codigo, qtd in pedidos:
    encontrado = False
    for p in produtos:
        if p["codigo"] == codigo:
            encontrado = True
            if p["estoque"] >= qtd:
                p["estoque"] -= qtd
                valor = qtd * p["preco"]
                valor_total += valor
                print(f"Pedido {codigo} ({p['nome']}, {qtd} unid) ATENDIDO - R$ {valor:.2f}")
            else:
                print(f"Pedido {codigo} ({p['nome']}, {qtd} unid) RECUSADO - estoque insuficiente")
            break
    if not encontrado:
        print(f"Pedido {codigo} RECUSADO - código inexistente")

print("\nEstoque final:")
for p in produtos:
    print(p["codigo"], "-", p["nome"], "-", p["estoque"], "unidades")

print(f"\nValor total dos pedidos atendidos: R$ {valor_total:.2f}")
```

```python
#15. Sistema de vendas e relatório final
produtos = [
    {"codigo": "P01", "nome": "Arroz", "preco": 5.50, "estoque": 10, "vendido": 0, "valor_vendido": 0},
    {"codigo": "P02", "nome": "Feijão", "preco": 7.20, "estoque": 8, "vendido": 0, "valor_vendido": 0},
    {"codigo": "P03", "nome": "Macarrão", "preco": 4.00, "estoque": 15, "vendido": 0, "valor_vendido": 0},
    {"codigo": "P04", "nome": "Óleo", "preco": 6.30, "estoque": 12, "vendido": 0, "valor_vendido": 0},
    {"codigo": "P05", "nome": "Sal", "preco": 2.50, "estoque": 20, "vendido": 0, "valor_vendido": 0},
]

historico = []
qtd_vendas = 0
unidades_vendidas = 0
valor_total = 0

while True:
    codigo = input("Código do produto (0 encerra): ")
    if codigo == "0":
        break
    qtd = int(input("Quantidade: "))
    encontrado = False
    for p in produtos:
        if p["codigo"] == codigo:
            encontrado = True
            if p["estoque"] >= qtd:
                p["estoque"] -= qtd
                p["vendido"] += qtd
                valor = qtd * p["preco"]
                p["valor_vendido"] += valor
                qtd_vendas += 1
                unidades_vendidas += qtd
                valor_total += valor
                historico.append((codigo, qtd, valor))
                print("Venda registrada.")
            else:
                print("Venda recusada, estoque insuficiente.")
            break
    if not encontrado:
        print("Código inexistente.")

print("\nEstoque restante:")
for p in produtos:
    print(p["codigo"], "-", p["nome"], "-", p["estoque"], "unidades")

print("\nUnidades vendidas por produto:")
for p in produtos:
    print(p["codigo"], "-", p["nome"], "-", p["vendido"], "unidades")

print("\nValor vendido por produto:")
for p in produtos:
    print(p["codigo"], "-", p["nome"], "- R$", f"{p['valor_vendido']:.2f}")

maior_valor = produtos[0]["valor_vendido"]
for p in produtos:
    if p["valor_vendido"] > maior_valor:
        maior_valor = p["valor_vendido"]

print("\nProduto(s) de maior valor em vendas:")
for p in produtos:
    if p["valor_vendido"] == maior_valor and maior_valor > 0:
        print(p["nome"], "- R$", f"{p['valor_vendido']:.2f}")

print("\nProdutos sem venda:")
for p in produtos:
    if p["vendido"] == 0:
        print(p["nome"])

print("\nHistórico de vendas:")
for h in historico:
    print(h)

resumo = (qtd_vendas, unidades_vendidas, valor_total)
print("\nResumo final:", resumo)
```
