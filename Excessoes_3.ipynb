```python
#1
notas = []

for i in range(5):
    while True:
        try:
            nota = float(input(f"Digite a {i+1}ª nota: "))

            if not 0 <= nota <= 10:
                print("A nota deve estar entre 0 e 10.")
                continue

            notas.append(nota)
            break

        except ValueError:
            print("Entrada inválida. Digite um número.")

soma = 0
for nota in notas:
    soma += nota

media = soma / len(notas)

acima_ou_igual = 0
for nota in notas:
    if nota >= media:
        acima_ou_igual += 1

print("Notas:", notas)
print("Média:", media)
print("Notas >= média:", acima_ou_igual)
```

```python
# Questão 2 · Coordenada representada por tupla
while True:
    try:
        x = float(input("Digite a coordenada x: "))
        y = float(input("Digite a coordenada y: "))
        break
    except ValueError:
        print("Entrada inválida! Por favor, digite apenas números.")

ponto = (x, y)

if ponto == (0.0, 0.0):
    print(f"O ponto {ponto} está localizado na Origem.")
elif ponto[0] == 0.0:
    print(f"O ponto {ponto} está sobre o eixo Y.")
elif ponto[1] == 0.0:
    print(f"O ponto {ponto} está sobre o eixo X.")
elif ponto[0] > 0 and ponto[1] > 0:
    print(f"O ponto {ponto} está no 1º Quadrante.")
elif ponto[0] < 0 and ponto[1] > 0:
    print(f"O ponto {ponto} está no 2º Quadrante.")
elif ponto[0] < 0 and ponto[1] < 0:
    print(f"O ponto {ponto} está no 3º Quadrante.")
else:
    print(f"O ponto {ponto} está no 4º Quadrante.")
```

```python
# Questão 3 · Cadastro simples de produto
nome = input ("Digite o nome do produto: ")

while True:
  try:
    preco = float(input("Digite o preço do produto: "))
    if preco < 0:
      print ("Entrada inválida! Digite o valor do produto positivo.")
      continue
    break
  except ValueError:
    print ("Entrada inválida, digite valores numéricos.")

while True:
  try:
    quantidade = int (input("Digite a quantidade de produtos: "))
    if quantidade < 0:
      print ("Entrada inválida! Digite a quantidade positiva .")
      continue
    break
  except ValueError:
    print ("Entrada inválida, digite valores numéricos.")

produto = {
    "nome" : nome,
    "preco" : preco,
    "quantidade" : quantidade
}

valor_total = produto ["preco"] * produto ["quantidade"]

print(f"Cadastro do produto: {produto}")
print(f"Valor total em estoque: R$ {valor_total:.2f}")
```

```python
# Questão 4 · Lista de números: pares e ímpares
numeros = []

for i in range (1,11):
  while True:
    try:
      num = int (input(f"Digite o {i}º número: "))
      numeros.append (num)
      break
    except ValueError:
      print("Entrada inválida! Por favor, digite um número inteiro.")
pares = []
impares =[]

for num in numeros:
  if num % 2 == 0:
    pares.append (num)
  else:
    impares.append (num)
print(f"Lista completa: {numeros}")
print(f"Lista de pares: {pares}")
print(f"Lista de ímpares: {impares}")
```

```python
# Questão 5 · Maior e menor da lista
numeros = []

for i in range (1,9):
  while True:
    try:
      num = int (input(f"Digite o {i}º número inteiro: "))
      numeros.append (num)
      break
    except ValueError:
      print("Entrada inválida! Por favor, digite um número inteiro.")

maior = numeros [0]
menor = numeros [0]
pos_maior = 1
pos_menor = 1

for i in range(1, len(numeros)):
  if numeros[i] > maior:
    maior = numeros[i]
    pos_maior = i +1
  elif numeros [i] < menor:
    menor = numeros [i]
    pos_menor = i + 1

print (f"Lista completa: {numeros}")
print(f"Maior valor: {maior} (na posição {pos_maior})")
print(f"Menor valor: {menor} (na posição {pos_menor})")
```

```python
# Questão 6 · Nomes sem repetição
nomes = []

while True:

  nome = input("Digite um nome (FIM para encerrar): ")

  if nome.upper() == "FIM":
    break

  elif nome == "":
    print ("O nome não pode estar vazio. Tente novamente.")
    continue

  if nome not in nomes:
    nomes.append(nome)

  else:
    print ("Este nome ja foi cadastrado")

print(f"Lista de nomes diferentes: {nomes}")
print(f"Quantidade de nomes diferentes cadastrados: {len(nomes)}")
```

```python
# Questão 8 · Estatísticas de uma tupla

numeros = []

while len(numeros) < 6:
    try:
        num = int(input(f"Digite o {len(numeros) + 1}º número: "))
        numeros.append(num)
    except ValueError:
        print("Entrada inválida! Digite um número inteiro.")

tupla = tuple(numeros)

qtd_5 = tupla.count(5)
pares = sum(1 for x in tupla if x % 2 == 0)

print(f"\nTupla gerada: {tupla}")

if qtd_5 > 0:
    print(f"O número 5 apareceu {qtd_5} vez(es), na posição {tupla.index(5)}.")
else:
    print("O número 5 não apareceu nenhuma vez.")

if pares > 0:
    print(f"Total de valores pares: {pares}.")
else:
    print("Nenhum valor par foi encontrado.")

print("\n--- Teste ---")
teste = (5, 2, 5, 8, 3, 4)
print(f"Tupla de teste: {teste}")
print(f"Quantidade de 5: {teste.count(5)}")
print(f"Pares no teste: {sum(1 for x in teste if x % 2 == 0)}")
```

```python
# Questão 9 · Dicionário de aluno

while True:
  try:
    nota1 = float (input("Digite a primeira nota: "))
    break
  except ValueError:
    print ("Entrada inválida!!")


while True:
  try:
    nota2 = float (input("Digite a segunda nota: "))
    break
  except ValueError:
    print ("Entrada inválida!!")
aluno = {
    "nome" : input ("Digite o nome do aluno: "),
    "turma" : input ("Digite a turma: "),
    "nota1" : nota1,
    "nota2" : nota2
}

aluno ["media"] = (aluno["nota1"] + aluno["nota2"] )/2

if aluno["media"] >= 7:
  aluno ["situacao"] = "Aprovado"
elif aluno["media"] >= 5:
    aluno["situacao"] = "Recuperação"
else:
    aluno["situacao"] = "Reprovado"

print("\nDados do aluno: ")
for chave, valor in aluno.items():
    print(f"{chave}: {valor}")
```

```python
# Questão 10 · Consulta de preços
# Criação do dicionário com pelo menos 5 produtos e seus preços
precos_produtos = {
    "maçã": 4.50,
    "banana": 3.20,
    "laranja": 5.00,
    "leite": 4.80,
    "pão": 10.00
}

print("=== Sistema de Consulta de Preços ===")
print("Produtos cadastrados:", ", ".join(precos_produtos.keys()))

# Loop com while para consultas interativas
while True:
    try:
        # Leitura do nome do produto digitado pelo usuário
        produto = input("\nDigite o nome do produto para consultar o preço (ou 'sair' para encerrar): ").strip().lower()

        # Condicional para encerramento do programa
        if produto == 'sair':
            print("Encerrando a consulta de preços. Até logo!")
            break

        # Validação de regras com if/elif/else
        if not produto:
            print("Erro: O nome do produto não pode estar vazio. Tente novamente.")
        elif produto in precos_produtos:
            # Produto encontrado
            print(f"O preço do produto '{produto}' é R$ {precos_produtos[produto]:.2f}.")
        else:
            # Produto não cadastrado (mensagem apropriada)
            print(f"Aviso: O produto '{produto}' não está cadastrado no sistema.")

    except ValueError:
        # Tratamento de exceções conforme exigido
        print("Erro: Valor inválido inserido. Tente novamente.")
```

```python
# Questão 11 • Frequência das palavras

frequencia_palavras = {}
contador = 0

print("=== Programa de Contagem de Frequência de Palavras ===")
print("Por favor, digite 8 palavras.")

# Loop para ler exatamente 8 palavras
while contador < 8:
    try:
        palavra = input(f"Digite a {contador + 1}ª palavra: ").strip().lower()

        # Validação para evitar entradas vazias
        if not palavra:
            print("Erro: A palavra não pode estar vazia. Tente novamente.")
            continue

        # Atualiza a contagem no dicionário sem utilizar .get()
        if palavra in frequencia_palavras:
            frequencia_palavras[palavra] += 1
        else:
            frequencia_palavras[palavra] = 1
        contador += 1

    except ValueError:
        print("Erro: Entrada inválida. Tente novamente.")

# Exibição do resultado final
print("\n=== Resultado Final • Frequência das Palavras ===")
for palavra, quantidade in frequencia_palavras.items():
    print(f"A palavra '{palavra}' foi digitada {quantidade} vez(es).")
```

```python
# Questão 12 · Estoque e atualização

estoque = {
    "teclado": 10,
    "mouse": 15,
    "monitor": 5,
    "headset": 8
}

print("=== Estoque Inicial ===")
for produto, qtd in estoque.items():
    print(f"{produto}: {qtd} unidades")

while True:
    try:
        produto = input("\nDigite o nome do produto (ou 'sair'): ").strip().lower()

        if produto == 'sair':
            break

        if not produto:
            print("Erro: O nome do produto não pode estar vazio. Tente novamente.")
            continue

        quantidade_vendida = int(input("Digite a quantidade vendida: "))

        if produto in estoque:
            if estoque[produto] >= quantidade_vendida > 0:
                estoque[produto] -= quantidade_vendida
                print(f"Venda registrada com sucesso! Novo estoque de '{produto}': {estoque[produto]}")
            elif quantidade_vendida <= 0:
                print("Erro: A quantidade vendida deve ser maior que zero.")
            else:
                print(f"Erro: Estoque insuficiente! Há apenas {estoque[produto]} unidades de '{produto}'.")
        else:
            print("Aviso: O produto não está cadastrado no estoque.")

    except ValueError:
        print("Erro: Digite um valor numérico válido para a quantidade.")

print("\n=== Estoque Final ===")
for produto, qtd in estoque.items():
    print(f"{produto}: {qtd} unidades")
```

```python
# Questão 13 · Cadastro de vários alunos
lista_alunos = []
contador = 0

print("=== Cadastro de 5 Alunos ===")
while contador < 5:
    nome = input(f"Digite o nome do {contador + 1}º aluno: ").strip()

    if not nome:
        print("Erro: O nome não pode estar vazio. Tente novamente.")
        continue

    while True:
        try:
            nota1 = float(input(f"Digite a 1ª nota de {nome}: "))
            if 0 <= nota1 <= 10:
                break
            else:
                print("Erro: A nota deve estar entre 0 e 10.")
        except ValueError:
            print("Erro: Valor numérico inválido. Tente novamente.")

    while True:
        try:
            nota2 = float(input(f"Digite a 2ª nota de {nome}: "))
            if 0 <= nota2 <= 10:
                break
            else:
                print("Erro: A nota deve estar entre 0 e 10.")
        except ValueError:
            print("Erro: Valor numérico inválido. Tente novamente.")

    media = (nota1 + nota2) / 2

    aluno = {
        "nome": nome,
        "notas": [nota1, nota2],
        "media": media
    }

    lista_alunos.append(aluno)
    contador += 1

print("\n=== Resultado Final • Alunos e Médias ===")
for aluno in lista_alunos:
    print(f"Aluno: {aluno['nome']} | Média: {aluno['media']:.2f}")
```

```python
# Questão 14 · Catálogo com código imutável
# Lista contendo 5 produtos em formato de tuplas: (codigo, nome, preco)
catalogo = [
    (1, "Notebook", 3500.00),
    (2, "Mouse", 80.00),
    (3, "Teclado", 150.00),
    (4, "Monitor", 1200.00),
    (5, "Headset", 250.00)
]

print("=== Consulta de Catálogo ===")

while True:
    try:
        codigo_busca = int(input("\nDigite o código do produto (ou 0 para sair): "))

        if codigo_busca == 0:
            print("Encerrando o programa. Até logo!")
            break

        encontrado = False

        for produto in catalogo:
            codigo, nome, preco = produto
            if codigo == codigo_busca:
                print(f"Produto encontrado -> Nome: {nome} | Preço: R$ {preco:.2f}")
                encontrado = True
                break

        if not encontrado:
            print("Aviso: O código informado não foi encontrado no catálogo.")

    except ValueError:
        print("Erro: Digite um valor numérico válido para o código.")
```

```python
# Questão 15 · Mini sistema de estoque

estoque_lista = []
contador = 0

print("=== Cadastro de Estoque (5 Itens) ===")
while contador < 5:
    try:
        codigo = int(input(f"\nDigite o código do {contador + 1}º item: "))
        nome = input(f"Digite o nome do {contador + 1}º item: ").strip()

        if not nome:
            print("Erro: O nome não pode estar vazio. Tente novamente.")
            continue

        quantidade = int(input(f"Digite a quantidade em estoque de {nome}: "))
        if quantidade < 0:
            print("Erro: A quantidade não pode ser negativa. Tente novamente.")
            continue

        preco = float(input(f"Digite o preço unitário de {nome}: "))
        if preco < 0:
            print("Erro: O preço não pode ser negativo. Tente novamente.")
            continue

        item = {
            "codigo": codigo,
            "nome": nome,
            "quantidade": quantidade,
            "preco": preco
        }

        estoque_lista.append(item)
        contador += 1

    except ValueError:
        print("Erro: Entrada numérica inválida. Tente novamente.")

# (a) Mostrar todos os itens
print("\n=== (a) Lista de Todos os Itens ===")
for item in estoque_lista:
    print(f"Código: {item['codigo']} | Nome: {item['nome']} | Qtd: {item['quantidade']} | Preço: R$ {item['preco']:.2f}")

# (b) Calcular o valor total do estoque e (c) identificar o item de maior valor
valor_total_estoque = 0.0
maior_valor_item = -1
item_mais_valioso = None

for item in estoque_lista:
    valor_item = item["quantidade"] * item["preco"]
    valor_total_estoque += valor_item

    if valor_item > maior_valor_item:
        maior_valor_item = valor_item
        item_mais_valioso = item

print(f"\n=== (b) Valor Total do Estoque ===")
print(f"R$ {valor_total_estoque:.2f}")

print(f"\n=== (c) Item de Maior Valor em Estoque ===")
if item_mais_valioso:
    print(f"Produto: {item_mais_valioso['nome']} | Valor Total em Estoque: R$ {maior_valor_item:.2f}")

# (d) Criar tupla resumo contendo (quantidade_de_itens_cadastrados, valor_total_estoque)
resumo_tupla = (len(estoque_lista), valor_total_estoque)

print(f"\n=== (d) Tupla Resumo ===")
print(resumo_tupla)
```
