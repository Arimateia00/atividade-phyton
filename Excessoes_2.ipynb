```python
# Questão 1 · Maior e menor entre vários valores
while True:
  try:
    primeiro = int (input("Digite o 1º número: "))
    break
  except ValueError:
    print ("Erro. Digite número válido.")
maior = primeiro
menor = primeiro

for i in range (5):
  while True:
    try:
      numero = int (input(f"Digite o {i + 2}º número:"))
      break
    except ValueError:
       print ("Erro. Digite número válido.")
  if numero > maior:
      maior = numero
  if numero < menor:
      menor = numero
print ("Maior:", maior)
print ("Menor:", menor)
```

```python
# Questão 2 · Estatísticas de uma turma
soma = 0
aprovados = 0
abaixo = 0

for i in range(5):
    while True:
        try:
            nota = float(input(f"Nota do aluno {i+1}: "))

            if not 0 <= nota <= 10:
                print("Nota deve estar entre 0 e 10.")
                continue # Volta ao início do ciclo while para pedir a nota novamente

            break # Sai do ciclo while se a nota for válida

        except ValueError:
            print("Digite uma nota numérica válida.")

    soma += nota

    if nota >= 7:
        aprovados += 1
    else:
        abaixo += 1

media = soma / 5

print("\n--- Estatísticas da Turma ---")
print(f"Média: {media:.2f}")
print("Notas >= 7:", aprovados)
print("Notas < 7:", abaixo)
```

```python
# 4 · Estacionamento

while True:
  try:
    horas = int (input("Digite a quantidade de horas que você ficou no estacionamento: "))
    if horas < 0:
      print("Erro: A quantidade de horas não pode ser negativa. Tente novamente.")
    else:
      break
  except ValueError:
    print ("Entrada inválida!!")
if horas <= 2:
  valor = 8
elif horas <= 5:
  valor = 8 + 3 * (horas - 2)
else:
  valor = 17 + 2 * (horas - 5)
print(f"Tempo estacionado: {horas} hora(s)")
print(f"Valor a pagar: R$ {valor:.2f}")
```

```python
# 5 · Máquina de venda de ingresso
total_inserido = 0
ingresso = 25
while total_inserido < ingresso:
  while True:
    try:
      valor = float (input("Digite o valor a inserir: "))
      if valor <= 0:
        print ("Erro, o valor tem que ser maior que zero.")
      else:
        break
    except ValueError:
            print("Entrada inválida. Digite um número inteiro.")
  total_inserido = total_inserido + valor
  if total_inserido < ingresso:
    falta = ingresso - total_inserido
    print (f"Falta {falta:.2f} reais")
troco = total_inserido - ingresso
print("\n--- Compra Concluída ---")
print (f"O valor total inserido foi de {total_inserido} reais.")
print (f"O valor do troco foi de {troco} reias.")
```

```python
#6 · Maior valor e sua posição
while True:
    try:
       primeiro = int (input("Digite o 1° número: "))
       break

    except ValueError:
       print ("Entrada inválida. Tente novamente! ")

maior = primeiro
posicao = 1
for i in range (7):
    while True:
       try:
          numero = int (input(f"Digite o {i + 2}° número: "))
          break
       except ValueError:
          print ("Erro! Tente novamente. ")
    # O 'if' deve ficar aqui fora do while, mas dentro do for!
    if numero > maior:
       maior = numero
       posicao = i + 2
print (f"O maior número é o {maior} na posição {posicao}. ")
```

```python
#7 · Segundo maior valor
while True:
   try:
      primeiro = int (input("Digite o 1° número: "))
      segundo = int (input ("Digite o 2° número: "))
      break
   except ValueError:
      print ("Entrada inválida! Digite valores numéricos. ")

if primeiro > segundo:
    maior = primeiro
    segundo_maior = segundo
else:
    maior = segundo
    segundo_maior = primeiro

for i in range (6):
   while True:
      try:
         numero = int (input (f"Digite o {i + 3}° número: "))
         break
      except ValueError:
         print ("Entrada inválida! Digite valores numéricos: ")
   if numero > maior:
      segundo_maior = maior
      maior = numero
   elif numero > segundo_maior:
      segundo_maior = numero
print (f"O maior número é o {maior} e o segundo maior é o {segundo_maior}. ")
```

```python
#8 · Detectando repetição consecutiva
tem_repeticao = False
while True:
     try:
        anterior = int (input ("Digite o 1° número: "))
        break
     except ValueError:
        print ("Entrada inválida. Tente novamente! ")
for i in range (7):
   while True:
      try:
         numero = int (input(f"Digite o {i + 2}° número: "))
         break
      except ValueError:
         print ("Entrada inválida. Tente novamente! ")
   if numero == anterior:
      tem_repeticao = True
      anterior = numero
if tem_repeticao:
    print ("Existem valores consecutivos iguais! ")
else:
    print ("Não existem valores consecutivos iguais!" )
```

```python
#9 · Controle de saldo
saldo = 500.0

print(f"Saldo inicial: R$ {saldo:.2f}\n")

for i in range(1, 7):
    print(f"--- Operação {i} de 6 ---")

    tipo = 0
    while True:
        try:
            tipo = int(input("Digite o tipo da operação (1 = Depósito, 2 = Saque): "))
            if tipo in [1, 2]:
                break
            else:
                print("Opção inválida! Digite 1 para Depósito ou 2 para Saque.")
        except ValueError:
            print("Entrada inválida! Digite apenas números inteiros (1 ou 2).")

    valor = 0.0
    while True:
        try:
            valor = float(input("Digite o valor da operação: R$ "))
            if valor > 0:
                break
            else:
                print("O valor deve ser maior que zero.")
        except ValueError:
            print("Entrada inválida! Digite um valor numérico válido.")

    if tipo == 1:
        saldo += valor
        print(f"Depósito de R$ {valor:.2f} realizado com sucesso!")
    elif tipo == 2:
        if valor <= saldo:
            saldo -= valor
            print(f"Saque de R$ {valor:.2f} realizado com sucesso!")
        else:
            print("Saldo insuficiente")

    print(f"Saldo atual: R$ {saldo:.2f}\n")

print(f"Saldo final: R$ {saldo:.2f}")
```

```python
# 11 · Número primo
while True:
  try:
    n = int (input("Digite um número inteiro positivo: "))
    if n > 1:
      break
    print ("O número precisa ser maior que 1")
  except ValueError:
    print ("Entrada inválida. Digite valores numéricos! ")
primo = True

for i in range (2,n):
  if n % i == 0:
    primo = False
    break

if primo:
  print (f"O número {n} é primo ")
else:
  print (f"O número {n} não é primo")
```

```python
# 12 · Divisores de um número
while True:
  try:
    n =int (input ("Digite um número para saber os divisores dele: "))
    if n > 0:
      break
  except ValueError:
    print ("Entrada inválida! Digite valores numéricos. ")
quantidade = 0
for i in range (1, n + 1):
  if n % i == 0:
    print (i)
    quantidade += 1

print (f"O número {n} possui {quantidade} divisores! ")
```

```python
# 13 · Números primos em um intervalo
while True:
  try:
    A = int (input("Digite o primeiro número: "))
    if A > 0:
      break
    print ("O valor deve ser maior que 1.")
  except ValueError:
    print ("Entrada inválida! Digite valores númericos. ")

while True:
  try:
    B = int (input("Digite o segundo número: "))
    if A < B:
      break
    print ("O segundo valor deve ser maior que o primeiro1.")
  except ValueError:
    print ("Erro. Tente novamente! ")

total_primos = 0
print(f"\nPrimos entre {A} e {B}:")

for num in range (A, B + 1 ):
  if num > 1:
    for i in range (2, num):
      if num % i == 0:
        break
    # O 'else' do for só executa se o loop terminou SEM dar break (ou seja, é primo)
    else:
      print (num)
      total_primos += 1
print(f"\nForam encontrados {total_primos} número(s) primo(s).")
```

```python
# Questão 14
numero_secreto = 37
tentativas = 5

print("=== Jogo do Número Secreto ===")

while tentativas > 0:
  try:
    numero = int (input("Digite o número secreto: "))

  except ValueError:
    print ("Entrada inválida. Digite a senha com números!! ")
    continue
  if numero == numero_secreto:
    print ("Parabéns, você acertou!!")
    break
  elif numero > numero_secreto:
    print ("Tente um número menor!")
  else:
    print ("Tente um número maior!")

  tentativas -= 1

if tentativas == 0:
  print(f"\nFim do jogo! As tentativas esgotaram. O número secreto era {numero_secreto}.")
```

```python
# Questão 15
saldo = 200.0
entradas = 0
retiradas = 0
recusadas = 0
total_entrou = 0.0
total_saiu = 0.0

while True:
    while True:
        try:
            codigo = int(input("Código (0=Sair, 1=Entrada, 2=Retirada): "))
            if codigo in [0, 1, 2]:
                break
            print("Código inválido.")
        except ValueError:
            print("Digite um número inteiro.")

    if codigo == 0:
        break

    while True:
        try:
            valor = float(input("Valor: R$ "))
            if valor > 0:
                break
            print("O valor deve ser maior que zero.")
        except ValueError:
            print("Digite um valor numérico.")

    if codigo == 1:
        saldo += valor
        total_entrou += valor
        entradas += 1
    elif codigo == 2:
        if valor <= saldo:
            saldo -= valor
            total_saiu += valor
            retiradas += 1
        else:
            recusadas += 1
            print("Saldo insuficiente")

print(f"\nSaldo final: R$ {saldo:.2f}")
print(f"Quantidade de entradas: {entradas}")
print(f"Retiradas realizadas: {retiradas}")
print(f"Retiradas recusadas: {recusadas}")
print(f"Total que entrou: R$ {total_entrou:.2f}")
print(f"Total que saiu: R$ {total_saiu:.2f}")
```
