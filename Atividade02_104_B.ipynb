```python
#4.Estacionamento
hrs = int (input("Digite a quantidade de horas: "))
if hrs <= 2 :
  valor_final = 8
  print("O valor a ser pago é: ", valor_final)
elif hrs <= 5 :
  valor_final = 8 + (hrs - 2) * 3
  print ("O valor a ser pago é: ", valor_final)
else:
    valor_final = 17 + (hrs - 5) * 2
    print ("O valor final a ser pago é: ", valor_final)
```

```python
#5. Máquina de venda de ingresso
ingresso = 25
total = 0
while total < ingresso:
  valor = int (input("Digite um valor inteiro: "))
  total = total + valor
  troco = total - ingresso
print ("O total inserido foi de:",total,"reais")
print ("O troco foi de:",troco , "reais")
```

```python
#6. Maior valor e sua posição
maior = int (input("Digite um número inteiro: "))
posicao = 1
for i in range (1,6):
  num = int (input("Digite um número inteiro: "))
  if num > maior:
    maior = num
    posicao = i
print ("Maior valor:", maior)
print ("Posição:", posicao)
```

```python
#7. Segundo maior valor
maior = int (input("Digite números inteiros distintos: "))
segundo_maior = int (input("Digite números inteiros distintos: "))

if segundo_maior > maior:
  maior, segundo_maior = segundo_maior, maior

for i in range (1,7):
  num = int (input("Digite números inteiros distintos: "))
  if num > maior:
    segundo_maior = maior
    maior = num
  elif num > segundo_maior :
    segundo_maior = num

print("O maior número é:",maior)
print ("O segundo maior número é:",segundo_maior)
```

```python
#8. Detectando repetição consecutiva
anterior = int (input("Digite um número: "))
repetiu = False
for i in range (7):
  num =int (input("Digite um número: "))
  if num == anterior:
    repetiu = True
  anterior = num
if repetiu:
  print("existem valores consecutivos iguais. ")
else:
  print ("não existem valores consecutivos iguais. ")
```

```python
#9. Controle de saldo
saldo = 500.00
for i in range (6):
  num = int (input("Digite o número da operação: (1 = depósito; 2 = saque)"))
  valor = float (input("Digite o valor: "))
  if num == 1:
    saldo = saldo + valor
  elif num == 2:
    if valor > saldo:
      print ("Saldo Insuficiente")
    else:
      saldo = saldo - valor
  else:
    print ("Numero Inválido")
print ("Saldo final: ", saldo)
```

```python
#11. Número primo
n = int (input("Digite um número inteiro positivo(maior que 1): "))
if n > 1:
  primo = True
  for d in range (2,n):
    if n % d == 0:
      primo = False
  if primo:
    print (n, "é primo")
  else:
    print (n,"não é primo")
else:
  print ("Número inválido!!")
```

```python
#12. Divisores de um número
n = int (input("Digite um número inteiro e positivo: "))
if n > 0:
  qtd_divisores = 0
  for d in range (1,n):
    if n % d == 0:
      print ("Os divisores de",n, "é:", d)
      qtd_divisores += 1
  print ("Numero de divisores: ",qtd_divisores)
else:
  print ("Número Inválido")
```

```python
#13. Números primos em um intervalo
A = int (input("Digite um número inteiro positivo: "))
B = int (input("Digite um número inteiro positivo(maior que anterior): "))
contador = 0
if B > A:
  for i in range (A, B + 1):
    primo = True
    for d in range (2,i):
      if i % d == 0:
        primo = False
    if primo and i > 1:
      print (i,"é primo")
      contador += 1
  print ("A quantidade de primos encontrados: ", contador)
else:
  print ("Números Inválidos")
```

```python
#14. Jogo com limite de tentativas
nsecreto = 37
tentativas = 5
for i in range (tentativas):
  senha = int (input("Digite a senha:"))
  if senha == nsecreto:
    print("Parabéns, você acertou a senha!!")
    break
    if tentativas > 0:
      tentativas -= 1
  elif senha > nsecreto:
    print ("Erro, tente um número menor!!")
  elif senha < nsecreto:
    print ("Erro, tente um número maior!!")
  tentativas -= 1
  print ("Tentativas restantes: ", tentativas)

if tentativas == 0:
  print ("Acabaram as tentativas!!")
```

```python
#15. Desafio final - caixa de atendimento
caixa = 200
qtd_entradas = 0
qtd_retiradas = 0
qtd_retiradas_recusadas = 0
total_entradas = 0
total_saidas = 0
while True:
  codigo = int (input("Digite o código(0=encerrar, 1=entrada, 2=retirada): "))
  if codigo == 0:
    break
  elif codigo == 1:
    valor = float (input("Digite o valor da entrada: "))
    caixa += valor
    qtd_entradas += 1
    total_entradas += valor
  elif codigo == 2:
    valor = float (input("Digite o valor o valor da retirada: "))
    if valor <= caixa:
      caixa -= valor
      qtd_retiradas += 1
      total_saidas += valor
    else:
      print ("Retirada recusada! Saldo Insuficiente!!")
      qtd_retiradas_recusadas += 1
  else:
    print ("Código Inválido!!")
print ("/n Relatório final")
print ("Saldo final",caixa)
print ("Quantidade de entradas realizadas:", qtd_entradas)
print ("Quantidade de retiradas recusadas:",qtd_retiradas_recusadas)
print ("Total de entradas: ", total_entradas)
print ("Total saídas:",total_saidas)
```
