```python
#8. Estatísticas de uma tupla
numeros = []
for i in range (6):
  num = int (input("Digite um número: "))
  numeros.append (num)

tupla = tuple (numeros)

qtd_5 = tupla.count (5)

if 5 in tupla:
  pos_5 = tupla.index (5) + 1
else:
  print ("Não apareceu")
pares = 0
for n in tupla:
  if n % 2 == 0:
    pares += 1
print ("Tupla",tupla)
print ("O número 5 apareceu", qtd_5, "vez(es)")
print ("Primeira ocorrência do número 5:", pos_5)
print ("Quantidade de pares:", pares)
```

```python
#9. Dicionário de aluno
aluno = {
    "nome " : input ("Digite o nome do aluno: "),
    "turma" : input ("Digite a turma:"),
    "nota1" : float (input("Digite a primeira nota: ")),
    "nota2" : float (input("Digite a segunda nota: "))
    }
aluno ["media"] = (aluno ["nota1"] + aluno ["nota2"]) / 2

if aluno ["media"] >= 7:
  aluno ["situacao"] = ("Aprovado")
elif aluno ["media"] >= 5:
  aluno ["situacao"] = ("Recuperação")
else:
  aluno ["situacao"]= ("Reprovado")

print ("\nDados do aluno: ")
for chave, valor in aluno.items ():
  print (f"{chave}: {valor}")
```

```python
#11. Frequência das palavras
palavras = []
for i in range (8):
  palavra = input ("Digite uma palavra: ")
  palavras.append (palavra)

frequencia = {}

for palavra in palavras:
  if palavra in frequencia:
    frequencia [palavra] += 1
  else:
    frequencia [palavra] = 1
print ("\nFrequência das palavras:",frequencia)
for palavra, contagem in frequencia.items ():
  print (f"{palavra} : {contagem} ")
```

```python
#12. Estoque e atualização
estoque = {
    "arroz" : 4,
    "feijao" : 1,
    "oleo" : 12,
    "farinha" : 9

    }
produto = input ("Digite o nome do produto: ")
quantidade = int (input("Digite a quantidade: "))

if produto in estoque:
  if estoque [produto] >= quantidade:
    estoque [produto] -= quantidade
    print (f"Venda realizada! {quantidade} unidade(s) do {produto} vendida(s).")
  else:
    print (f"Estoque Insuficiente! Disponível: {estoque[produto]}")
else:
  print ("Produto não encontrado no estoque!")
print ("\nEstoque final:")
for item, qtd in estoque.items ():
  print (f"{item} : {qtd}")
```

```python
#13. Cadastro de vários alunos
turma = []
for i in range (5):
  print (f"\nAluno {i+1}º")
  nome = input ("Digite o nome do aluno: ")
  n1 = float (input("Digite a primeira nota do aluno: "))
  n2 = float (input("Digite a segunda nota do aluno: "))

  media = (n1 + n2) / 2

  aluno = {
      "nome" : nome,
      "n1" : n1,
      "n2": n2,
      "media" : media
      }
  turma.append(aluno)

print ("\nMédia das turmas")
for aluno in turma:
 print(f"{aluno['nome']} - Média: {aluno['media']:.2f}")
```

```python
#14. Catálogo com código imutável
catalogo = [
    (101, "Arroz", 12.50),
    (102, "Feijão", 8.90),
    (103, "Macarrão", 5.75),
    (104, "Óleo", 9.20),
    (105, "Açúcar", 4.30)
]

codigo_busca = int(input("Digite o código do produto: "))

encontrado = False
for produto in catalogo:
    if produto[0] == codigo_busca:
        print(produto[1], "custa R$", f"{produto[2]:.2f}")
        encontrado = True
        break

if not encontrado:
    print("Código não encontrado.")
```
