```python
#Questão 1 Positivo, negativo ou zero
while True:
  try:
    num = int(input("Digite um número: "))
    break
  #Tratar entradas de texto caso o usuário digite.
  except ValueError:
    print ("Digite um número inteiro: ")
if num > 0:
  print("Positivo")
elif num < 0:
  print ("Negativo")
else:
  print ("Zero")
```

```python
# Questão 3 · Somando cinco números
soma = 0
for i in range(5):
  while True:
    try:
      numero = int(input("Digite um número: "))
      soma += numero
      break
    # Intercepta dados inválidos (como letras).
    except ValueError:
      print("Entrada inválida. Tente novamente.")
print("Soma:", soma)
```

```python
# Questão 4 · Par ou ímpar
while True:
  try:
    numero = int(input("Digite um inteiro: "))
    break
  # Trata o ValueError disparado quando o input não pode ser convertido para número inteiro
  except ValueError:
      print("Use apenas um número inteiro.")
if numero % 2 == 0:
  print("Par")
else:
  print("Ímpar")
```

```python
# Questão 5 · Situação do aluno
notas = []
for i in range(2):
  while True:
    try:
      nota = float(input(f"Digite a {i+1}ª nota: "))
      if 0 <= nota <= 10:
        notas.append(nota)
        break
      print("A nota deve estar entre 0 e 10.")
    # Trata o ValueError gerado quando a string digitada não é compatível com float()
    except ValueError:
      print("Digite uma nota numérica válida.")
media = sum(notas) / 2
if media >= 7:
  print("Aprovado")
elif media >= 5:
  print("Recuperação")
else:
  print("Reprovado")
print("Média:", media)
```

```python
# 7 · Contagem crescente
while True:
    try:
        n = int(input("Digite um número inteiro: "))
        break
    except ValueError:
        # Trata entrada de letras ou valores inválidos; a estratégia repete o pedido para buscar apenas números inteiros.
        print("Digite um número inteiro válido.")

if n > 0:
    for i in range(1, n + 1):
        print(i)
```

```python
# 8 · Soma de 1 até N
soma = 0
while True:
  try:
    n = int (input("Digite um número inteiro: "))

    if n <= 0:
      print ("Digite um número inteiro positivo. ")
      continue

    for i in range (1, n + 1):
      soma = soma + i
    print (f"A soma de {n} até 1 é {soma}.")
    break

  #Trata entrada de letras. A estratégia faz buscar apenas números.
  except ValueError:
    print ("Erro! Digite um número inteiro. ")
```

```python
# 9 · Contando números pares
pares = 0
contador = 1
while True:
    for i in range(10):
        try:
            num = int(input(f"Digite o {contador}º número: "))

            if num % 2 == 0:
                pares += 1
            # Se der erro, o contador não muda e a mesma posição é pedida novamente!
            contador += 1

          #Trata se a entrada não for um número inteiro; a estratégia mantém a posição e avisa o utilizador.
        except ValueError:
            print("Erro! Digite um número inteiro.")
    break

print(f"Foram digitados {pares} números pares.")
```

```python
# 10 · Soma apenas dos positivos
soma_positivos = 0
while True:
  try:
    for i in range (8):
      numero = int (input("Digite um número inteiro positivo: "))

      if numero >= 0:
        soma_positivos = soma_positivos + numero
    break

  #Trata entradas que não sejam inteiros
  except ValueError:
    print ("Erro! ")
print ("A soma dos números postivos é:", soma_positivos)
```

```python
# 11 · Maior valor digitado
maior = None
contador = 1
while contador <= 6:
  try:
    num = int (input(f"Digite o {contador}º número: "))

    if maior is None or num > maior:
      maior = num
    contador += 1

  # Trata entradas não inteiras mantendo os valores já aceitos
  except ValueError:
        print("Erro! Digite um número inteiro válido.")

print (f"O maior valor digitado foi: {maior}")
```

```python
# 12 · Tabuada
while True:
  try:
    num = int (input("Digite um número inteiro para ver a tabuada:"))
    break

  # Trata o erro caso a entrada seja texto ou número decimal
  except ValueError:
    print ("Erro! Digite um número inteiro positivo.")

print (f"Tabuada do {num}:")

for i in range (1,11):
  resultado = num * i
  print (f"Tabuada do {num} * {i} = {resultado}")
```

```python
# 13 · Tentativas de senha
senha_correta = 1234
senha = 0
while senha != senha_correta:
  try:
      senha = int (input("Digite a senha: "))

      if senha != senha_correta:
        print ("Senha incorreta. Tente novamente.")

  # Trata entradas não inteiras (letras, decimais, etc.)
  except ValueError:
    print ("Digite a senha com números inteiros.")

print ("Acesso permitido")
```

```python
# 14 · Número secreto
num_secreto = 7
tentativas = 0
entradas_invalidas = 0
while True:
  try:
    num = int (input("Digite a senha: "))
    tentativas += 1
    if num < num_secreto:
      print ("Senha incorreta. Tente um número maior.")

    elif num > num_secreto:
      print ("Senha incorreta. Tente um número menor.")

    else:
      print ("Parabéns. Você acertou o número secreto.")
      break
  #Caso o usuário digite algo difrente do número vai contar como entrada inválida.
  except ValueError:
    print ("Digite o número secreto com número(s) inteiro.")
    entradas_invalidas += 1

print (f"Você acertou o número secreto com {tentativas} tentativa(s) e {entradas_invalidas} entradas invalidas(s).")
```

```python
# 15 · Analisando uma sequência
quantidade = 0
positivos = 0
negativos = 0
soma = 0

while True:
    try:
        num = int(input("Digite um número inteiro (0 para encerrar): "))

        if num == 0:
            break

        quantidade += 1
        soma += num

        if num > 0:
            positivos += 1
        elif num < 0:
            negativos += 1

    except ValueError:
        # Critério: Entrada inválida não encerra nem altera os resultados
        print("Erro! Digite um número inteiro válido.")

print("\n--- Resultados da Análise ---")
print(f"Quantidade total: {quantidade}")
print(f"Positivos: {positivos}")
print(f"Negativos: {negativos}")
print(f"Soma: {soma}")
```
