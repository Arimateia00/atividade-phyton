```python
#4. Par ou ímpar
num = int (input("Digite um número: "))
if num  % 2 == 0 :
  print ("O número é par")
else:
  print ("O número é impar")
```

```python
#5. Situação do aluno. Leia duas notas de um aluno e calcule sua média.
nota1 = float (input("Digite a primeira nota do aluno: "))
nota2 = float (input("Digite a segunda nota do aluno: "))
media = (nota1 + nota2) / 2
if media >= 7 :
  print ("Aprovado", media)
elif media >= 5:
  print ("Recuperação",media)
else:
  print ("Reprovado", media)
```

```python
#6. Maior entre três números
num1 = int (input("Digite o primeiro número: "))
num2 = int (input("Digite o segundo número: "))
num3 = int (input("Digite o terceiro número: "))
if num1 == num2 == num3:
  print ("Todos são iguais")
elif num1 >= num2 and num1 >= num3:
  print ("O primeiro número é o maior:", num1 )
elif num2 >= num1 and num2 >= num3:
  print ("O segundo número é o maior:", num2)
else:
  print ("O terceiro número é o maior: ",num3)
```

```python
#7. Contagem crescente.Leia um número inteiro positivo N e mostre todos os números de 1 até N. Exemplo para N = 5:
N = int (input("Digite um número inteiro positivo: "))
if N > 0:
    for i in range(1, N + 1):
        print (i)
else:
    print ("Número Inválido!! ")
```

```python
#8. Somando de 1 até N. Leia um número inteiro positivo N e calcule a soma de todos os números de 1 até N.
N = int (input("Digite um número inteiro positivo: "))
if N > 0:
  soma = 0
  for i in range(1, N + 1):
        soma += i
  print ("Soma de 1 até ",N, "=",soma)
else:
    print ("Número Inválido!! ")
```

```python
#9. Contando números pares
pares = 0
for i in range (10):
  num = int (input(f"Digite o {i+1}º número: "))
  if num % 2 == 0:
    pares = pares + 1
print ("Foram digitados",pares , "números pares")
```

```python
#10. Soma apenas dos positivos. Leia 8 números inteiros.
positivo = 0
for i in range (8):
  num = int (input(f"Digite o {i+1}º número: "))
  if num > 0:
    positivo += num
print ("Soma dos positivos: ",positivo)
```

```python
#11. Maior valor digitado
maior = int (input("Digite o 1º número: "))
for i in range (2,7):
  num = int (input(f"Digite o {i+1}º número: "))
  if num > maior :
    maior += num
print ("O maior número é:", maior)
```

```python
#12. Tabuada. Leia um número inteiro e mostre sua tabuada de 1 até 10.
num = int (input ("Digite um número: "))
print (f"Tabuada do {num}: ")
for i in range (1,11):
  resultado = num * i
  print (f"{num} x {i} = {resultado}")
```

```python
#13. Tentativas de senha
senhacorreta = 1234
senha = int (input("Digite a senha de 4 dígitos: "))
while senha != senhacorreta:
  print ("Senha incorreta.Tente novamente ")
  senha = int (input("Digite a senha de 4 dígitos: "))
print ("Acesso Permitido! ")
```

```python
#14. Número secreto
num_secreto = 7
tentativas = 0

num = int (input ("Digite um número: "))
tentativas += 1

while num != num_secreto:
  if num < num_secreto:
    print ("Tente um número maior")
  else:
      print ("Tente um número menor")
  num = int (input ("Digite novamente: "))
  tentativas += 1
print ("Parabéns! Você acertou.")
print ("Número de tentativas: ", tentativas)
```

```python
#15. Analisando uma sequência de números
quantidade = 0
positivos = 0
negativos = 0
soma = 0

num = int(input("Digite um número (0 para encerrar): "))
while num != 0 :
  quantidade += 1
  soma += num
  if num > 0:
    positivos += 1
  if num < 0:
    negativos += 1
  num = int(input("Digite outro número (0 para encerrar): "))
print ("Quantidade de números digitados: ", quantidade)
print ("Quantidade de números positivos: ",positivos)
print ("Quantidade de números negativos: ",negativos)
print ("Soma de todos os números: ",soma)
```
